```python
img = cv2.imread('/content/drive/My Drive/IP/Images/lena.jpg')
print(img.shape)
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
#plt.imshow(gimg1, cmap='gray')
cv2_imshow(gimg)
```

```python
outimg = cv2.blur(gimg,(7,7))
cv2_imshow(outimg)
```

```python
outimg = cv2.GaussianBlur(gimg,(7,7), 5)
cv2_imshow(outimg)
```

```python
outimg = cv2.medianBlur(gimg, 7)
cv2_imshow(outimg)
```

```python
outImage = gimg.copy()
#blurImage = cv2.GaussianBlur(outImage, (7,7), 5)
blurImage = cv2.medianBlur(gimg, 7)

x = 200 + 70
y = 200 + 65
size = 15

for h in range(y, y+size):
  for w in range(x, x+size):
    #outImage[h][w] = 255
    outImage[h][w] = blurImage[h][w]

cv2_imshow(outImage)
```

```python
img = cv2.imread('/content/drive/My Drive/IP/Images/lena256.jpg')
print(img.shape)
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
#plt.imshow(gimg1, cmap='gray')
cv2_imshow(gimg)
```

```python
outimg = cv2.GaussianBlur(gimg,(7,7), 2)
cv2_imshow(outimg)
```

```python
edgeImage = cv2.addWeighted(gimg, 1, outimg, -1, 0)
cv2_imshow(edgeImage)
```

```python
UnsharpImage = cv2.addWeighted(gimg, 1, edgeImage, 2, 0)
cv2_imshow(UnsharpImage)
```
