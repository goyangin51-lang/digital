```python
from google.colab.patches import cv2_imshow
from google.colab import drive
import numpy as np
import cv2
import matplotlib.pyplot as plt
drive.mount('/content/drive')
```

```python
img = cv2.imread('/content/drive/My Drive/IP/Images/lena.jpg')
print(img.shape)
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
#plt.imshow(gimg, cmap='gray')
cv2_imshow(gimg)
```

```python
outImg = np.zeros(gimg.shape, dtype=np.uint8)
#cv2_imshow(outImg)
```

```python
scaling = 1.5

for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    y = int(h/scaling)
    x = int(w/scaling)
    if y < gimg.shape[0] and x < gimg.shape[1]:
      outImg[y,x] = gimg[h,w]

cv2_imshow(outImg)
```

```python
scaling = 3

for y in range(gimg.shape[0]):
  for x in range(gimg.shape[1]):
    h = int(y/scaling)
    w = int(x/scaling)
    if h < gimg.shape[0] and w < gimg.shape[1]:
      outImg[y,x] = gimg[h,w]

cv2_imshow(outImg)
```

```python
#outImg = cv2.resize(gimg, None, fx=3, fy=3, interpolation = cv2.INTER_NEAREST)
#outImg = cv2.resize(gimg, None, fx=3, fy=3, interpolation = cv2.INTER_LINEAR)
outImg = cv2.resize(gimg, None, fx=3, fy=3, interpolation = cv2.INTER_CUBIC)
cv2_imshow(outImg)
```

```python
img = cv2.imread('/content/drive/My Drive/IP/Images/lena256.jpg')
print(img.shape)
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
#plt.imshow(gimg, cmap='gray')
cv2_imshow(gimg)
```

```python
height, width = gimg.shape
print(height, width)
```

```python
M = cv2.getRotationMatrix2D((width/2.0, height/2.0), 45, 1)
print(M)
```

```python
outImg = cv2.warpAffine(gimg, M, (width, height))
cv2_imshow(outImg)
```

```python
M = np.float32([[1, 0, 30], [0, 1, 20]])
outImg = cv2.warpAffine(gimg, M, (width, height))
cv2_imshow(outImg)
```
